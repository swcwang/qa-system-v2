```python
from qanda_module.config import setup_system
import re

# Setup
helpers, qa_chain, config = setup_system("config.yaml")

# Test query
query = "What did panelists say about climate change?"
print(f"Query: {query}")
```

```python
# ====================
# STEP 1: What does retrieval return?
# ====================
print("\n" + "="*50)
print("STEP 1: RETRIEVAL ANALYSIS")
print("="*50)

docs = qa_chain.retriever._get_relevant_documents(query)
print(f"Retrieved {len(docs)} documents")

# Sample first 3 docs
for i in range(min(3, len(docs))):
    doc = docs[i]
    print(f"\n--- DOC {i} ---")
    print(f"Metadata keys: {list(doc.metadata.keys())}")
    print(f"Episode ID: {doc.metadata.get('episode_id', 'MISSING')}")
    print(f"Content preview: {doc.page_content[:200]}...")
```

```python
# ====================
# STEP 2: What does AI see?
# ====================
print("\n" + "="*50)
print("STEP 2: AI CONTEXT ANALYSIS")
print("="*50)

context_text = "\n\n".join(doc.page_content for doc in docs)
print(f"Total context length: {len(context_text)} chars")
print(f"\nFirst 300 chars:")
print(context_text[:300])
print(f"\nLast 300 chars:")
print(context_text[-300:])

# Check for episode patterns
ep_patterns = re.findall(r'Ep \d+', context_text)
episode_id_patterns = re.findall(r'Episode ID: \d+', context_text)
print(f"\nEpisode patterns found:")
print(f"'Ep XXX' format: {ep_patterns[:10]}...")  # First 10
print(f"'Episode ID: XXX' format: {episode_id_patterns[:10]}...")  # First 10
```
